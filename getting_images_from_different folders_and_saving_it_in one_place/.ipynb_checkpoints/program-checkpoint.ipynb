```python
import glob
import os
from PIL import Image,ImageFile
```

```python
ImageFile.LOAD_TRUNCATED_IMAGES = True
```

```python
for name in glob.glob('C:/Users/User/Downloads/sd_card/*'):
    name_splitted = name.split('\\')
    folder_name = name_splitted[-1]
    is_folder = folder_name.split('.')

    if len(is_folder) == 1:
        os.chdir(name)
        if len(os.listdir())!=0:
            dir_list = os.listdir()
            for content in dir_list:
                is_image = content.split('.')
                #print(is_image)
                if(len(is_image) > 1):
                    if is_image[-1] in ['png','jpg','jpeg','JPG']:
                        print(content)
                        is_image = '.'.join(is_image)
                        og_pic_path = os.getcwd()+ '\\' +content
                        picture = Image.open(og_pic_path)
                        #print(":")
                        os.chdir('C:/Users/User/Downloads/images')
                        picture = picture.save(is_image)
                        os.remove(og_pic_path)
                        os.chdir(name)
```
