```python
import PyPDF2
```

```python
# Separate pdf file paths, that are supposed to be merged
pdf_files_list = [
    "C:/Users/10677003/OneDrive - LTIMindtree/Documents/visa/certifications/certifications.pdf",
    "C:/Users/10677003/OneDrive - LTIMindtree/Documents/visa/certifications/databricks.pdf"
]
# Final merged pdf path
final_pdf_path = "./merged_pdf.pdf"
```

```python
# Pdf writer object
final_pdf_writer_obj = PyPDF2.PdfWriter()
```

## With Pdf Merger

```python
# Pdf merger object
merger_obj = PyPDF2.PdfMerger()
# Appending files to the merger object
for pdf in pdf_files_list:
    merger_obj.append(pdf)
# Writing the merged object
merger_obj.write(final_pdf_path)
# Close the pdf merger object
merger_obj.close()
```

## Without Pdf Merger
This gives you more control over your pdfs

```python
for pdf_file in pdf_files_list:
    # Open current pdf file
    # Here we are not using `with open()`, because we were getting ValueError..
    # seems like if reader object closes once the file is closed.
    pdf_file_obj = open(pdf_file, "rb")
    # Pdf reader object
    document_file_reader_obj = PyPDF2.PdfReader(pdf_file_obj)

    # Getting total number of pages in a pdf for `for` loop
    total_pages = len(document_file_reader_obj.pages)
    # Iterating over all the pages and adding it to the final writer object
    for pageNum in range(total_pages):
        doc_page_obj = document_file_reader_obj.pages[pageNum]
        final_pdf_writer_obj.add_page(doc_page_obj)
    
    # Closing the pdf file object
    pdf_file_obj.close()
```

```python
# Writing the pdf writer object
with open(final_pdf_path, "wb") as f:
    final_pdf_writer_obj.write(f)
```
